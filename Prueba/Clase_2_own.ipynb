#Las listas pueden ser modificadas

```python
lista_repetida = ["Historia", "Economía", "Antropología", "RSP", "Derecho", "Historia", "Economía", "Antropología", "RSP", "Derecho"]
print(lista_repetida)
```

```python
lista_repetida[2] = "Finanzas"
print(lista_repetida)
```

```python
lista_1 = [1,2,3,4,5]
lista_2 = lista_1
print(lista_1)
print(lista_2)
```

```python
lista_1[3] = 10
print(lista_1)
print(lista_2)
```

```python
print(id(lista_1))
print(id(lista_2))
```

```python
lista_2 = lista_1.copy()
lista_1[3] = 4
print(lista_1)
print(lista_2)
```

```python
print(id(lista_1))
print(id(lista_2))
```
